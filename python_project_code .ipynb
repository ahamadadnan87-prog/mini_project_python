```python
import numpy as np
import pandas as pd

# Initial hotel data
data = {
    "Room_No": [101, 102, 103, 104, 105],
    "Room_Type": ["Single", "Double", "Deluxe",
                  "Single", "Deluxe"],
    "Price": [1000, 1500, 2500, 1000, 2500],
    "Customer": ["-", "-", "-", "-", "-"],
    "Status": ["Available", "Available",
               "Available", "Available", "Available"]
}

hotel = pd.DataFrame(data)


# Display all rooms
def display_rooms():
    print("\n--- HOTEL ROOM DETAILS ---")
    print(hotel.to_string(index=False))


# Add room
def add_room():
    global hotel

    room = int(input("Enter room number: "))

    if room in hotel["Room_No"].values:
        print("Room already exists!")
        return

    rtype = input("Enter room type: ")
    price = float(input("Enter room price: "))

    new_room = pd.DataFrame([{
        "Room_No": room,
        "Room_Type": rtype,
        "Price": price,
        "Customer": "-",
        "Status": "Available"
    }])

    hotel = pd.concat([hotel, new_room], ignore_index=True)
    print("Room added successfully!")


# Book room
def book_room():
    room = int(input("Enter room number: "))

    idx = hotel.index[hotel["Room_No"] == room]

    if len(idx) == 0:
        print("Room not found!")
        return

    i = idx[0]

    if hotel.at[i, "Status"] == "Occupied":
        print("Room is already occupied!")
        return

    name = input("Enter customer name: ")

    hotel.at[i, "Customer"] = name
    hotel.at[i, "Status"] = "Occupied"

    print("Room booked successfully!")


# Customer checkout
def checkout():
    room = int(input("Enter room number: "))

    idx = hotel.index[hotel["Room_No"] == room]

    if len(idx) == 0:
        print("Room not found!")
        return

    i = idx[0]

    if hotel.at[i, "Status"] != "Occupied":
        print("Room is not occupied!")
        return

    hotel.at[i, "Customer"] = "-"
    hotel.at[i, "Status"] = "Available"

    print("Checkout successful!")


# Search room
def search_room():
    room = int(input("Enter room number: "))

    result = hotel[hotel["Room_No"] == room]

    if result.empty:
        print("Room not found!")
    else:
        print(result.to_string(index=False))


# Hotel statistics
def statistics():
    total = len(hotel)
    available = np.sum(hotel["Status"] == "Available")
    occupied = np.sum(hotel["Status"] == "Occupied")

    print("\n--- HOTEL STATISTICS ---")
    print("Total Rooms:", total)
    print("Available Rooms:", available)
    print("Occupied Rooms:", occupied)

    print("Total Room Value:", np.sum(hotel["Price"]))


# Revenue calculation
def revenue():
    total = np.sum(
        hotel.loc[hotel["Status"] == "Occupied", "Price"]
    )

    print("Current occupied-room daily revenue:", total)


# Save data
def save_data():
    hotel.to_csv("hotel_data.csv", index=False)
    print("Data saved successfully!")


# Main menu
while True:

    print("\n========== HOTEL MANAGEMENT SYSTEM ==========")
    print("1. Display All Rooms")
    print("2. Add New Room")
    print("3. Book Room")
    print("4. Customer Checkout")
    print("5. Search Room")
    print("6. Hotel Statistics")
    print("7. Calculate Daily Revenue")
    print("8. Save Data")
    print("9. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_rooms()

    elif choice == "2":
        add_room()
    elif choice == "3":
        book_room()

    elif choice == "4":
        checkout()

    elif choice == "5":
        search_room()

    elif choice == "6":
        statistics()

    elif choice == "7":
        revenue()

    elif choice == "8":
        save_data()

    elif choice == "9":
        save_data()
        print("Thank you for using Hotel Management System!")
        break

    else:
        print("Invalid choice! Try again.")
```
